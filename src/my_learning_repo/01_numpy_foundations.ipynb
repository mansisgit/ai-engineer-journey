```python
import numpy as np

# 1D array (vector)
a = np.array([10, 20, 30])
print("a shape:", a.shape)  # prints (3,)

# 2D array (matrix with 2 rows, 3 columns)
b = np.array([[1, 2, 3], [4, 5, 6]])
print("b shape:", b.shape)  # prints (2, 3)
print("b dtype:", b.dtype)
```

```python
arr = np.array([[1,3,4],[5,7,7]])
print(arr[0,1])

b = np.array([[1, 2, 3], [4, 5, 6]])
print(b.shape)

# Create a column vector with shape (2, 1)
col = np.array([[100], [200]])
print("col shape:", col.shape)

# Broadcast a (2, 1) array across a (2, 3) array:
# The column stretches across all 3 columns
print("b + col:\n", b + col)
c = b+col
print(c.shape)
```

```python
a = np.array(10)

print(a.shape)
print(a.ndim)
# ()
```

```python
a = np.array([[2,3,4,5],
              [2,3,4,5],
              [6,7,8,9]])
print(a.shape)

b = np.array([2,2,6,2])
print(b.shape)

c = a+b
print(c.shape)

```
